```python
# 파일 입출력
# 파일 출력 -> 파일 입력
file = open("test.txt", "w", encoding="UTF-8")  # 파일 생성
file.write("안녕, 세계!\n") # 파일 입력
file.write("반가워")
```

```python
file.close()    # 파일 닫기
```

```python
# with open()
class MyResource:
    def __enter__(self):
        print("__enter__ 호출")

        return self

    # with 구문 종료 or 예외 발생 시 호출
    # 예외 발생 시 exc_type : 예외 클래스 종류
    #             exc_val : 예외 메세지
    #             exc_tb : TraceBack
    def __exit__(self, exc_type, exc_val, exc_tb):
        print("__exit__ 호출")
```

```python
with MyResource() as o:     # __enter__ 호출
    print(o)

    # 종료 시 __exit__ 호출
```

```python
# with 구문
# 파일 쓰기 작업
# 파일에서 작성하는 위치(파일 포인터)가 가장 앞으로 이동해서 내용을 대체
with open("test.txt", "w", encoding="UTF-8") as fp:
    fp.write("Hello!\n")      # 기존 내용 지우고 새로 작성(덮어씌우기)
```

```python
# 파일 포인터를 이동하지 않고 누적해서 작성 (mode: a)
with open("test.txt", "a", encoding="UTF-8") as fp:
    fp.write("Nice to meet you")
```

```python
# 읽기 함수
# read
with open("test.txt", "r", encoding="UTF-8") as fp:
    text = fp.read()
    print(text)
```

```python
# 읽기 함수
# readline
# 라인을 모두 읽으면 None
with open("test.txt", "r", encoding="UTF-8") as fp:
    text = fp.readline()
    print(text)
```

```python
# 읽기 함수
# readline
# while문으로 반복 출력
with open("test.txt", "r", encoding="UTF-8") as fp:
    while(text := fp.readline()):   # := 대입과 판단을 동시에 하는 연산자
        print(text)
```

```python
# 읽기 함수
# readlines
# 리스트에 담아서 반환
with open("test.txt", "r", encoding="UTF-8") as fp:
    lines = fp.readlines()
    print(lines)
```

```python
# 읽기 함수
# 파일 포인터를 반복해서 readlines과 동일한 효과 가능
with open("test.txt", "r", encoding="UTF-8") as fp:
    for line in fp:
        print(line)
```

```python
# csv 형식 예시
import csv

with open("member.csv", "w", encoding="UTF-8", newline="") as fp:
    writer = csv.writer(fp)
    writer.writerow(["이름", "나이", "전화번호"])
    writer.writerow(["고영희", 22, "010-1111-0000"])
```

```python
with open("member.csv", "r", encoding="UTF-8") as fp:
    reader = csv.reader(fp)
    members = list(reader)

    print(members)
```

```python
# JSON 형식
import json

# dict -> JSON 파일로 저장 (직렬화)
user = {
    "name": "고영희",
    "age": 22,
    "mobile": "010-1111-0000"
}

with open("user.json", "w", encoding="UTF-8") as fp:
    json.dump(user, fp, indent=4, ensure_ascii=False)   # 각 항목마다 4칸씩 들여쓰기
```

```python
# JSON -> dict 으로 읽기 (역직렬화)
with open("user.json", "r", encoding="UTF-8")as fp:
    user2 = json.load(fp)

    print(user2, "type:", type(user2))
```
