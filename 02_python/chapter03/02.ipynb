```python
# 딕셔너리
# 생성
user = {
        "email": "test@test.org",
        "password": 1234,
        "name": "김철수"
        }
```

```python
# 딕셔너리
# 추가
user['mobile'] = "010-0000-0000"

print(user)
```

```python
# 딕셔너리
# 조회 - key 이름
print(user["name"])
```

```python
# 딕셔너리
# 조회

print(user.keys())      # .keys(), 모든 키 view
print(user.values())    # .values(), 모든 값 view
print(user.items())     # .items(), (키, 값) 목록 view
```

```python
# 딕셔너리
# 예외 방지 작동
#user['moblie']      # KeyError
moblie = user.get("moblie")                # 키가 없으면 None 반환 

print(type(moblie))

moblie = user.get("moblie", "값 없음")      # 키 이름으로 조회 되지 않으면 설정된 기본값으로 대체
print(moblie)
```

```python
# 딕셔너리
# 키 삭제
del user['mobile']
user
```

```python
# 집합(set) 자료구조
a = {10, 10, 10, 20, 30}

print(a)
# print(a[0])     # 위치에 대한 계산 불가
```

```python
# 집합(set) 연산
a = {10, 20, 30, 40}
b = {30, 40, 50, 60}

print(f"합집합: {a | b}")
print(f"교집합: {a & b}")
print(f"차집합: {a - b}")
print(f"대칭차 집합: {a ^ b}")

# 딕셔너리는 정렬이 되지 않아, 리스트나 튜플로 전환해서 정렬하기!
```

```python
# 집합 연산
# 집합은 내부 해시 기반 구조 덕분에 리스트나 튜플에 비해 포함 여부를 확인하는 연산 속도가 압도적으로 빠르다

print('email' in user)
print('email' not in user)
```
