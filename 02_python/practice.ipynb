```python
# 8장 1강 3-1
# 튜터님 풀이

class Hero:
    def __init__(self, name):
        self.name = name
        self.hp = 100

    def take_damage(self, damage):
        
        minus_damage = damage if self.hp >= damage else self.hp     # 삼항조건, 조건이 맞으면 앞, 아니면 뒤
        self.hp -= minus_damage
        print(f"남은 체력: {self.hp}")

        if self.hp <= 0:
            print(f"{self.name} 캐릭터가 쓰러졌습니다.")
```

```python
hero = Hero("고영희")
hero.take_damage(90)
hero.take_damage(20)
```

```python
# 8장 2강 2-1
# 튜터님 풀이

class DoorLock:
    def __init__(self, password):
        if not self.validate(password):
            return                          # 조건이 맞지 않으면 실행하지 않음
        self.__password = str(password)

    def get_password(self):
        return self.__password

    def set_password(self, password):
        if not self.validate(password):
            return
        
        self.__password = str(password)
        print(f"정상 변경되었습니다")

    def validate(self, password):
        result = len(str(password)) > 4
        if not result:
            print(f"경고: 비밀번호는 4자리 이상이어야 합니다")

        return result

door = DoorLock(0000)
door.__password = 'asdfg'

print(door.get_password())

# 값 변경 가능한 편법
door.DoorLock_password = 'asdfgg'
print(door.get_password())
```

```python
# 8장 2강 3-1
# 튜터님 풀이

class User:
    def __init__(self, name):
        self.name = name

class VIPUser(User):
    def __init__(self, name, point):
        super().__init__(name)
        self.point = point

user = VIPUser("영희", 20)
print(user.name, user.point)
```
