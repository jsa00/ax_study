```python
# mean(), median()
import statistics

nums = [10, 15, 20, 8, 7, 1000000]

print(f"평균: {statistics.mean(nums)}, 중앙값: {statistics.median(nums)}")
```

```python
# 리스트 컴프리헨션
nums = [*range(1, 11)]

nums2 = [x**2 for x in nums]    
print(nums2)

nums3 = [x for x in nums if x % 2 == 1]
print(nums3)
```

```python
fruits = ['사과', '오렌지', '멜론', '망고']

num = 1
for fruit in fruits:
    print(num, fruit)
    num += 1
```

```python
# enumerate() 예시
for item in enumerate(fruits):      # (순서 번호, 요소)
    print(item)
```

```python
# enumerate()
# 튜플 언패킹 - 기본
for item in enumerate(fruits):      # (순서 번호, 요소)
    idx = item[0]
    fruit = item[1]
    print(idx, fruit)
```

```python
# enumerate()
# 튜플 언패킹 - 간략
for i, fruit in enumerate(fruits):      
    print(i, fruit)
```

```python
# enumerate()
# 튜플 언패킹 + 인덱스 시작점 변경
for i, fruit in enumerate(fruits, start=1):      
    print(i, fruit)
```

```python
# 서로 다른 리스트에 있는 값을 쌍을 지어 튜플로 묶기
persons = ["김철수", "고영희", "김민수"]
fruits = ['사과', '오렌지', '멜론']

items = []
for i, person in enumerate(persons):
    items.append((person, fruits[i]))

items
```

```python
# zip()
items = zip(persons, fruits)

list(items)
```

```python
# sorted()
import random

nums = [*range(1, 11)]
random.shuffle(nums)        # 랜덤하게 섞어줌

nums
```

```python
# sorted()
# sort()
nums.sort()

print("nums:", nums)
```

```python
# sorted()
# sorted() 사용 시 변수에 담기 필수, 새로운 리스트를 만들어서 넣기 때문에
nums_sorted = sorted(nums)

print("nums_sorted:", nums_sorted)
```

```python
# min(), max()
print(min(nums), max(nums))
```
