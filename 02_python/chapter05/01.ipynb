```python
# 함수 정의
def calc(x):            # 입력, x=매개변수
    y = x * 2 + 1       # 처리

    return y            # 출력(반환)

result = calc(12)       # 함수 호출, 12=인수
print(result)
```

```python
# 함수 개체 참조
calc2 = calc

calc2(20)
```

```python
# 함수 저장 메모리 확인
calc is calc2
```

```python
# 함수 전역변수와 지역변수
def add(num1, num2):
    num3 = 30

    result = num1 + num2 + num3     # 지역변수 : result

    return result

result = add(10, 20)                # 전역변수 : result
print(result)
```

```python
# 함수 실행 순서
def func(num):              # 1 : func()
    result = add(10, 20)    # 2 : add()
    result += num

    return result

result = func(40)
print(result)
```

```python
# 함수 참조 연결 끊김
calc = 10
calc2 = 20
calc, calc2
```

```python
# return
def add(num):
    if num == 999:
        print("종료")
        return

    result = num ** 2

    return result

add(999)
```

```python
# 위치 인수와 키워드 인수
def add(num1, num2):
    return num1 + num2

add(10, 20)                 # 위치 인수
add(num1 = 10, num2 = 20)   # 키워드 인수
add(num2 = 20, num1 = 10)   # 키워드 이름이 중요! 순서는 노상관
```

```python
# 위치 인수와 키워드 인수
# 위치 인수 먼저 할당 후, 키워드 인수 할당
def add(num1, num2, name, mobile):
    return f"{num1 + num2}, {name=}, {mobile=}"

add(10, 20, mobile="010-0000-0000", name="고영희")
# add(name="고영희", 10, 20, mobile="010-0000-0000")    # 오류
```

```python
# 기본값 매개변수
def add(num1, num2, name, mobile = "010"):          # 기본값 mobile = "010"
    return f"{num1 + num2}, {name=}, {mobile=}"

add(10, 20, "고영희")                               # 기본값 입력 생략 가능
```

```python
# 기본값 매개변수 오류
# 위치 인수 할당 후에 키워드 인수를 할당해야함
# def add(num1, num2, name = None, mobile):
#     return f"{num1 + num2}, {name=}, {mobile=}"

# add(10, 20, "010-0000-0000")
```

```python
def add(num1, num2, name = None):
    return f"{num1 + num2}, {name=}"

add(10, 20, "010-0000-0000")
```

```python
# 가변 인수와 키워드 가변 인수
def add(*nums):                 # *nums = 가변 인수
    print(nums, type(nums))

add(10, 20)
add(10, 20, 30)
add(10, 20, 30, 40)
```

```python
# 가변 인수와 키워드 가변 인수
def add(*nums):
    total = 0
    for num in nums:
        total += num

    return total

print(add(10, 20))
print(add(10, 20, 30))
print(add(10, 20, 30, 40))
```

```python
# 키워드 가변 인수(**kwargs)
def print_user_info(**info):
    print(info)

print_user_info(name="김철수", age=20)
```

```python
# 인수 + 키워드 가변 인수(**kwargs)
def print_user_info(name, age, **info):
    print("name:", name)
    print("age:", age)
    print(info)

print_user_info("고영희", 20, mobile="010-0000-0000")
```

```python
# 가변 인수 + 키워드 가변 인수
# 위치 맞춰서 입력 추가!
def print_user_info(name, age, *args, **kwargs):
    print("name:", name)
    print("age:", age)
    print(args)
    print(kwargs)

print_user_info("고영희", 20, 'A', 20, mobile="010-0000-0000")
```

```python
# return문
# 하나의 함수로 사칙연산 모두 반환받기
# return문은 반환값이 하나이며, 콤마(,)로 이을 경우 튜플로 묶임
def cale(num1, num2):
    return num1 + num2, num1 - num2, num1 * num2, num1 / num2

a, b, c, d = cale(10, 2)    # 언패킹
print(a, b, c, d)
```

```python
# 프레임 객체 직접 확인하기
import sys

def outer_func():
    x = 10
    
    def inner_func():
        y = 20
        
        # 1. 현재 실행 중인 프레임 객체 가져오기
        frame = sys._getframe()
        
        print(f"1. 현재 실행 함수: {frame.f_code.co_name}")
        print(f"2. 현재 지역 변수(f_locals): {frame.f_locals}")
        print(f"3. 전역 스코프 참조 여부: {'outer_func' in frame.f_globals}")
        
        # 2. f_back을 통해 부모 프레임(outer_func)으로 이동
        parent_frame = frame.f_back
        print(f"4. 부모 함수 이름: {parent_frame.f_code.co_name}")
        print(f"5. 부모 함수의 지역 변수: {parent_frame.f_locals}")

    inner_func()

outer_func()
```
