```python
# 클로저
# 고차 함수
def outer():
    num1 = 10

    def inner():
        num2 = 20
        print(num1 + num2)

    return inner()

inner = outer()     # num1은 inner에 의해 사용되어 힙 영역에 생성됨
inner               # 10 + 20
```

```python
# 클로저
# 고차 함수
def add(num1):
    def _add(num2):
        def __add(num3):
            return num1 + num2 + num3
        
        return __add
    return _add

result1 = add(10)(20)(30)
result1
```

```python
# 클로저
# 고차 함수
add10 = add(10)

add10(30)(40)
```

```python
# 클로저
# 고차 함수
add10_20 = add(10)(20)

add10_20(30)
```

```python
# 데코레이터
def factoriall(num):
    result = 1

    for i in range(1, num+1):
        result *= i

    return result

factoriall(10)
```

```python
# 데코리에터
# 팩토리얼
def factorial2(num) :
    if num < 1:
        return 1
    
    return num * factorial2(num -1)  # 재귀 함수: 같은 함수 내에서 같은 함수 호출 방법

factorial2(10)
```

```python
# 팩토리얼 실행 걸린 시간 측정
import time

start = time.time()

result = factoriall(10)
print(result)

end = time.time()
print(f"걸린 시간: {end - start}")
```

```python
# 재귀 함수 실행 걸린 시간 측정
start = time.time()                 # 공통 기능

result = factorial2(10)             # 핵심 기능
print(result)                    

end = time.time()                   # 공통 기능
print(f"걸린 시간: {end - start}")
```

```python
# 데코레이터
# 시간 측정 기능 함수화
import time

def time_checker(callback, num) :
    start = time.time()                 # 공통 기능

    result = callback(num)              # 핵심 기능 - 팩토리얼 연산 함수(factorial1, factorial2)

    end = time.time()                   # 공통 기능
    print(f"걸린 시간: {end - start}")

    return result

time_checker(factoriall, 10)
time_checker(factorial2, 10)
```

```python
# 데코레이터 패턴
# 파이썬이 제공하는 문법
import time

def time_check2(callback):              
    def wrapper(*args, **kwargs):           # time_check : 관례적 명칭, *args, **kwargs로 활용성 높임
        start = time.time()                 # 공통 기능

        result = callback(*args, **kwargs)  # 핵심 기능

        end = time.time()                   # 공통 기능
        print(f"걸린 시간: {end - start}")

        return result

    return wrapper
```

```python
# 알아서 감싸줌
@time_check2
def factorial2(num) :
    if num < 1:
        return 1
    
    return num * factorial2(num -1)

# 출력
factorial2(10)
```

```python
# 데코레이터 활용
@time_check2
def add(num1, num2):
    return num1 + num2

add(10, 20)
```

```python
# 가변 인수 - 위치 매개변수
a = [10, 20]
def minus(num1, num2):
    return num1 - num2

minus(*a)               # *변수명 : 가변 인수 하나씩 펼쳐서 하나씩 넣어줌
```
