```python
# 이터러블
nums = [10, 20, 30]

nums.__iter__       # __iter__ : 
```

```python
# 이터러블
for num in nums:
    print(num)
```

```python
# 이터레이터
fruits = ['Apple', 'Ornage', 'Melon', 'Banana']

it = iter(fruits)   # __iter__(...) 호출
```

```python
next(it)    # 요소를 반복하는데, 끝나면 오류남 -> 예외처리 필요
```

```python
# 제너레이터
def generater_nums():

    print("1번째")
    yield 10            # 함수 실행 중지

    print("2번째")
    yield 20

    print("3번째")
    yield 30
```

```python
# 제너레이터
gen = generater_nums()
gen
```

```python
next(gen)   # 꺼내올 때 마다 힙 영역에 해당 값 할당
```

```python
# 제너레이터 표현식
gen = (i for i in range(1, 101) if i % 2 == 0)
gen
```

```python
next(gen)
```

```python
for num in gen:
    print(num)
```
