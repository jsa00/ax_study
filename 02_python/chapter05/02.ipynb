```python
# 전역 변수와 지역 변수
c = 30                                  # 전역 변수

def outer():
    a = 10

    print("c:", c)                      # 전역 변수 출력

    def inner():        
        b = 20
        print(f"{a=}, {b=}, {c=}")      

    inner()

outer()
```

```python
# LEGB 스코프
a = 'Global'

def outer():
    a = 'Enclosing'
    print("outer:", a)

    def inner():
        a = 'Local'
        print("inner:", a)

    inner()
outer()
```

```python
# global 키워드
a = 'Global'                # 전역 변수

def outer():
    a = 'Enclosing'         # 지역 변수
    print("outer:", a)  

    def inner():
        print("inner:", a)  # 우선 순위로 outer() a 출력

    inner()
outer()
```

```python
# global 키워드
# 함수 내에서 전역 변수 찾기
a = 'Global'                # 전역 변수

def outer():
    a = 'Enclosing'         # 지역 변수
    print("outer:", a)      

    def inner():
        global a            # 전역 변수를 찾는 a
        print("inner:", a)  # 전역 변수 a 출력

    inner()
outer()
```

```python
# global 키워드
# 부작용 - 값 변경?
a = True                    # 전역 변수

def outer():
    a = 'Enclosing'         # 지역 변수
    print("outer:", a)      

    def inner():
        global a            # 전역 변수를 찾는 a
        a = not a           
        print("inner:", a)  # 전역 변수 a 출력

    inner()
outer()
```

```python
# 일급 함수
# 함수를 매개변수로 사용하기
def callback():
    print("callback()")

def func(a):
    print("fnuc()")
    a()

func(callback)
```

```python
# 일급 함수
# 콜백 함수
def callback2():
    print("callback()")

def func(callback):
    print("fnuc()")
    callback()

func(callback2)
```

```python
# 일급 함수
# return 키워드를 활용한 반환값으로 함수 내보내기
def outer(num1):
    def inner(num2):
        return num1 + num2

    return inner

a = outer(10)       # 항상 10을 더하는 새로운 함수 - num1이 inner 함수에 속박(Closure)
print(a(20))        # 10 + 20

b = outer(20)       # 항상 20을 더하는 새로운 함수
print(b(20))        # 20 + 20

print(a(30))        # 10 + 30
print(b(30))        # 20 + 30
```

```python
# 일급 함수
# 콜백 함수
def change_numbers(*args, callback):
    nums = list(args)

    for i in range(len(nums)):
        nums[i] = callback(nums[i])
    
    return nums

def square(num):
    return num ** 2

change_numbers(10, 20, 30, callback=square)
```

```python
# 일급 함수
# 콜백 함수 ↑
def plus(num):
    return num + 10

change_numbers(10, 20, 30, callback=plus)
```

```python
# 람다 함수 예시
square = lambda num: num ** 2

square(10)
```

```python
# 람다 함수
def change_numbers(callback, *args):
    nums = list(args)

    for i in range(len(nums)):
        nums[i] = callback(nums[i])
    
    return nums

print(change_numbers(lambda num: num ** 2, 10, 20, 30))
print(change_numbers(lambda x: x ** 2, 10, 20, 30))        # num을 x로 짧게 변환 가능
```

```python
# lambda
# map()

fruits = ['사과', '오렌지', '망고']

list(map(lambda x: f"**{x}**", fruits))
```

```python
# lambda + callback
# filter()
nums = [*range(1, 11)]   # * 하나씩 펼쳐줌 (1~10)

list(filter(lambda x: x % 2 == 1, nums))
```
