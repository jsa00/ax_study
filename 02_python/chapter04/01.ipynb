```python
# 단일 조건문 if 예시
num = 10

if num > 5:
    print("5보다 큼")
```

```python
# if-else 조건문
num = 10
if num == 10:
    print(f"10이다.")
else:
    print(f"10이 아니다.")
```

```python
# input() : 사용자 입력 받기
age = int(input())
print(age, type(age))
```

```python
# 다중 조건 분기 : if-elif-else
age = int(input())

if age >= 20:
    print("성인")
elif age >= 17:
    print("고등학생")
elif age >= 14:
    print("중등학생")
elif age >= 8:
    print("초등학생")
else:
    print("유치원생")
```

```python
# 복잡한 조건 처리(중첩)
# 입력 받은 숫자가 10 이상, 100 미만인지 확인
num = int(input())

if num >= 10:
    if num < 100:
        print(f"{num}은 10 이상, 100 미만이다.")
    else:
        print(f"{num}은 10이상, 100 미만이 아니다.")
```

```python
# 복잡한 조건 처리(중첩) - 논리 연산자 and 결합
num = int(input())

if num >= 10 and num < 100:
    print(f"{num}은 10 이상, 100 미만이다.")
else:
    print(f"{num}은 10이상, 100 미만이 아니다.")
```
