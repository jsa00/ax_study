1. asyncio.gather()

```python
import asyncio

async def work(title, seconds):
    await asyncio.sleep(seconds)
    print(title, "완료!")
```

```python
async def main():
    await work("작업A", 2)
    await work("작업B", 3)
    await work("작업C", 1)
```

```python
await main()
```

```python
# asyncio.gather()
import asyncio

async def work(title, seconds):
    await asyncio.sleep(seconds)
    return title + " 완료!"
```

```python
async def main():
    results = await asyncio.gather(work("작업A", 2), work("작업B", 3), work("작업C", 1))
    print(results)
```

```python
await main()
```
