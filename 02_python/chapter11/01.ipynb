# 11장 1강 : 동기(Sync) vs 비동기(Async)

```python
# 동기식 함수 (서브 루틴)
import time

def work(title, seconds):
    time.sleep(seconds)         # 주로 동기식 방식에서 사용
    print(title, "완료!")

start = time.perf_counter()     # 작업 시작 시간

work("A작업", 2)
work("B작업", 3)
work("C작업", 1)

end = time.perf_counter()       # 작업 종료 시간
print("걸린 시간:", end - start)
```

```python
# 비동기식 함수 (코루틴)
import asyncio
import time

async def async_work(title, seconds):   # 함수명 맨 앞에 async 추가
    asyncio.sleep(seconds)
    print(title, "완료!")

async def main():
    
    asyncio.gather(             # gather : 작업 여러개를 한꺼번에 실행
        async_work("A작업", 2),
        async_work("B작업", 3),
        async_work("C작업", 1)
    )

start = time.perf_counter()
await main()
end = time.perf_counter()
print("걸린 시간:", end - start)
```
