```python
# 4장 3강 1-1
order_database = [
    {"name": "김철수", "completed": True, "price": 60000},
    {"name": "이영희", "completed": False, "price": 35000},
    {"name": "박민수", "completed": True, "price": 25000},
    {"name": "최지영", "completed": True, "price": 85000},
    {"name": "정성민", "completed": False, "price": 90000}
]

order_total = 0
order_count = 0

for order in order_database:
    if order["completed"] == True:

        if order["price"] >= 50000:
            order_total += order["price"]
            order_count += 1

print(f"우수 고객 주문 건수: {order_count}건")
print(f"우수 고객 총 결제 금액: {order_total}건")
```

```python
# 4장 3강 2-1
fruit_sweetness = [12.5, 14.2, 11.8, 10.5, 15.0, 13.6, 11.2, 14.8, 12.0, 13.1]

sweetness_max = fruit_sweetness[0]
sweetness_min = fruit_sweetness[0]
sweetness_sum = 0

for fruit in fruit_sweetness:
    if fruit > sweetness_max:
        sweetness_max = fruit

    if fruit < sweetness_min:
        sweetness_min = fruit
    
    sweetness_sum += fruit

sweetness_total = len(fruit_sweetness)
sweetness_avg = sweetness_sum / sweetness_total

print(f"최고 과일 당도: {sweetness_max}")
print(f"최저 과일 당도: {sweetness_min}")
print(f"과일 평균 당도: {sweetness_avg}")
```

```python
# 4장 3강 3-1
import random

computer_number = random.randint(1, 51)
try_count = 0

print("=" * 22, "♣☆♣", "=" * 22)
print(f"보물을 1부터 50 사이의 숫자 방에 숨겨두었습니다.")
print("=" * 50)

while True:
    user_input = input("▶ 보물이 있는 방을 맞춰보세요 (1~50): ")
    guess = int(user_input)
    try_count += 1

    if guess < computer_number:
        print("↑ 보물은 더 큰 숫자의 방에 있습니다.")
    elif guess > computer_number:
        print("↓ 보물은 더 작은 숫자의 방에 있습니다.")
    else:
        print(f"☆ 축하합니다! {try_count}회 만에 보물을 찾았습니다. ☆")
        break
```

```python
# 5장 1강 3-1
def get_total_and_average(*args):
    if len(args) == 0:
        return 0

    total = sum(args)
    avg = total / len(args)
    return avg

result1 = get_total_and_average(90, 85, 95)
result2 = get_total_and_average(70, 80, 90, 100, 65)

print(f"3과목 평균 점수: {result1:.2f}")
print(f"5과목 평균 점수: {result2:.2f}")
```

```python
# 5장 3강 1-1
def message_store():
    storage = ""

    def add_message(message):
        nonlocal storage
        storage += f"{message}\n"

        return storage
    return add_message

my_storage = message_store()
print(my_storage("새로운 알림이 도착했습니다"))
print(my_storage("출석 체크가 완료되었습니다"))
```

```python
# 5장 3강 2-1
def deco_border(callback):
    def wrapper(*args, **kwargs):
        print("=" * 50)

        result = callback(*args, **kwargs)

        print("=" * 50)

        return result
    return wrapper

@deco_border
def show_welcome():
    print("환영합니다 스파르타님")

show_welcome()
```

```python
# 5장 3강 3-1
def auto_tracker(callback):
    def wrapper(*args, **kwargs):
        print(f"▶ [LOG] 호출된 함수명: {callback.__name__}")
        print(f"▶ [LOG] 주입된 인수 데이터: 위치인수={args}, 키워드인수={kwargs}")

        result = callback(*args, **kwargs)

        print(f"◀ [LOG] {callback.__name__} 함수의 정상 연산이 종료되었습니다.")

        return result
    return wrapper

@auto_tracker
def add_numbers(a, b):
    return a + b

print(add_numbers(5, b=8))
```
