```python
# 4장 3강 1-1
order_database = [
    {"name": "김철수", "completed": True, "price": 60000},
    {"name": "이영희", "completed": False, "price": 35000},
    {"name": "박민수", "completed": True, "price": 25000},
    {"name": "최지영", "completed": True, "price": 85000},
    {"name": "정성민", "completed": False, "price": 90000}
]

order_total = 0
order_count = 0

for order in order_database:
    if order["completed"] == True:

        if order["price"] >= 50000:
            order_total += order["price"]
            order_count += 1

print(f"우수 고객 주문 건수: {order_count}건")
print(f"우수 고객 총 결제 금액: {order_total}건")
```

```python
# 4장 3강 2-1
fruit_sweetness = [12.5, 14.2, 11.8, 10.5, 15.0, 13.6, 11.2, 14.8, 12.0, 13.1]

sweetness_max = fruit_sweetness[0]
sweetness_min = fruit_sweetness[0]
sweetness_sum = 0

for fruit in fruit_sweetness:
    if fruit > sweetness_max:
        sweetness_max = fruit

    if fruit < sweetness_min:
        sweetness_min = fruit
    
    sweetness_sum += fruit

sweetness_total = len(fruit_sweetness)
sweetness_avg = sweetness_sum / sweetness_total

print(f"최고 과일 당도: {sweetness_max}")
print(f"최저 과일 당도: {sweetness_min}")
print(f"과일 평균 당도: {sweetness_avg}")
```

```python
# 5장 1강 3-1
def get_total_and_average(*args):
    if len(args) == 0:
        return 0

    total = sum(args)
    avg = total / len(args)
    return avg

result1 = get_total_and_average(90, 85, 95)
result2 = get_total_and_average(70, 80, 90, 100, 65)

print(f"3과목 평균 점수: {result1:.2f}")
print(f"5과목 평균 점수: {result2:.2f}")
```

```python
# 5장 3강 1-1
def message_store():
    storage = ""

    def add_message(message):
        nonlocal storage
        storage += f"{message}\n"

        return storage
    return add_message

my_storage = message_store()
print(my_storage("새로운 알림이 도착했습니다"))
print(my_storage("출석 체크가 완료되었습니다"))
```

```python
# 5장 3강 2-1
def deco_border(callback):
    def wrapper(*args, **kwargs):
        print("=" * 50)

        result = callback(*args, **kwargs)

        print("=" * 50)

        return result
    return wrapper

@deco_border
def show_welcome():
    print("환영합니다 스파르타님")

show_welcome()
```

```python
# 5장 3강 3-1
def auto_tracker(callback):
    def wrapper(*args, **kwargs):
        print(f"▶ [LOG] 호출된 함수명: {callback.__name__}")
        print(f"▶ [LOG] 주입된 인수 데이터: 위치인수={args}, 키워드인수={kwargs}")

        result = callback(*args, **kwargs)

        print(f"◀ [LOG] {callback.__name__} 함수의 정상 연산이 종료되었습니다.")

        return result
    return wrapper

@auto_tracker
def add_numbers(a, b):
    return a + b

print(add_numbers(5, b=8))
```

```python
# 8장 1강 1-1
class book:
    def __init__(self, title, author):
        self.title = title
        self.author = author

    def info(self):
        print(f"책 제목: [{self.title}], 저자: [{self.author}]")

book1 = book("어린왕자", "앙투안 드 생텍쥐페리")
book2 = book("인간실격", "다자이 오사무")
book1.info()
book2.info()
```

```python
# 8장 1강 2-1
class SmartLamp:
    def __init__(self, model, is_on = False):
        self.model = model
        self.is_on = is_on

    def turn_on(self):
        self.is_on = True
        print(f"{self.model} 조명이 켜졌습니다.")

    def turn_off(self):
        self.is_on = False
        print(f"{self.model} 조명이 꺼졌습니다.")

on = SmartLamp("수면")
on.turn_on()
off = SmartLamp("수면")
off.turn_off()
```

```python
# 8장 1강 3-1
class Hero:
    def __init__(self, name):
        self.name = name
        self.hp = 100

    def take_damage(self, damage):
        self.hp -= damage
        if self.hp <= 0:
            print(f"[{self.name}] 캐릭터가 쓰러졌습니다.")
        else:
            print(f"HP가 {self.hp} 남았습니다.")
```

```python
info = Hero("전사")
info.take_damage(99)
```

```python
# 8장 2강 1-1
class Phone:
    def call(self):
        return f"통화를 시작합니다"
    
class SmartPhone(Phone):
    def call(self):
        return f"영상 통화를 시작합니다"

    def search_web(self):
        return f"인터넷 검색을 시작합니다"

calling = SmartPhone()
print(calling.call())
print(calling.search_web())
```
