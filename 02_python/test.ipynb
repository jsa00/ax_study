```python
# 3장 1강 1-1
fruits = ['apple', 'banana', 'cherry']
print(fruits)

fruits[1] = 'blueberry'
print(fruits)

colors = 'red', 'green', 'blue'
print(colors[-1])
```

```python
# 3장 1강 2-1
group1 = [1, 2]
group2 = [3, 4]

new_group = group1 + group2
print(new_group)

user_roles = ['admin', 'editor', 'member']
print('admin' in user_roles)
```

```python
# 3장 1강 3-1
numbers = [0, 1, 2, 3, 4, 5]

print(numbers[1:4])

first, *others = [10, 20, 30, 40]

print(first)
print(others)
```

```python
# 3장 2강 1-1
user_info = {
            "name": "Sparta",
            "age": 20
            }

user_info['age'] = 21
user_info['role'] = "admin"
user_none = user_info.get("phone", "010-0000-0000")

print(user_info)
print(user_none)
```

```python
# 3장 2강 2-1
raw_emails = ["a@sparta.com", "b@sparta.com", "a@sparta.com"]

unique_set = set(raw_emails)
refined_emails = list(unique_set)

print(refined_emails)
```

```python
# 3장 2강 3-1
menu = {
    "coffee": 4000,
    "tea": 4500
}

print(menu.keys())
print(menu.values())

set_a = {1, 2, 3}
set_b = {3, 4, 5}

print(set_a & set_b)
print(set_a | set_b)
print(set_a - set_b)
```
