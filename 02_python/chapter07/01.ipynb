```python
# 예외 처리 구문 예시
try:
    num = int(input())
    result = 10 / num

    print(result)

except:
    print("0이 아닌 정수를 입력하세요.")
    
else:
    print("정상처리")

finally:
    print("로깅 처리..")
```

```python
# 다중 예외 처리
try:
    num = int(input())
    result = 10 / num

    print(result)
    
except ZeroDivisionError:               # 에러명을 명시
    print("0이 아닌 정수를 입력하세요.")
except ValueError:
    print("정수만 입력하세요.")
```

```python
# 다중 예외 처리
# 튜플로 묶기
try:
    num = int(input())
    result = 10 / num

    print(result)
    
except (ZeroDivisionError, ValueError):
    print("0이 아닌 정수를 입력하세요.")
```

```python
# 다중 예외 처리
# 별칭 사용
try:
    num = int(input())
    result = 10 / num

    print(result)
    
except ZeroDivisionError as e:
    print(e)
```

```python
# 다중 예외 처리
# 자세한 에러 생략
try:
    num = int(input())
    result = 10 / num

    print(result)
    
except ZeroDivisionError:
    print("0이 아닌 정수를 입력하시요")
except Exception as e:                  # Exception은 가장 하단에 위치
    print(e)
```

```python
# traceback
# traceback.print_exc()
import traceback

try:
    result = 10 / 0
    print(result)
except:
    print("에러 발생!")
    traceback.print_exc()
```

```python
# traceback
# traceback.format_exc()
import traceback

try:
    result = 10 / 0
    print(result)
except:
    print("에러 발생!")
    message = traceback.format_exc()
    print("message", message)
```

```python
def login(user_id, user_pass):
    if user_id != "user01":
        print("아이디 불일치")
        print("회원 아이디 찾기 페이지 이동...")
        return
    
    if user_pass != "1234":
        print("비밀번호 불일치")
        print("회원 비밀번호 찾기 페이지 이동...")
        return

    print("로그인 성공!")
```

```python
# raise 예시
def login(user_id, user_pass):
    if user_id != "user01":
        raise ValueError("아이디 불일치 경고")      # 예외 발생 -> 코드 실행 흐름 중단
    
    if user_pass != 1234:
        raise ValueError("비밀번호 불일치 경고")

    print("로그인 성공!")
```

```python
# 회원 페이지
try:
    login("user01", 1234)
except ValueError as e:
    print(e)
    print("-> 회원 페이지로 이동...")
```

```python
# 관리자 페이지
try:
    login("user02", 1234)
except ValueError as e:
    print(e)
    print("-> 관리자 페이지로 이동...")
```

```python
# 사용자 정의 예외 처리 예시
class UserIdMismatchError(Exception):       # 예외 클래스
    def __str__(self):
        return "아이디가 불일치 합니다"

class UserPassMismatchError(Exception):
    def __str__(self):
            return "비밀번호가 불일치 합니다"
```

```python
def login(user_id, user_pass):
    if user_id != "user01":
        raise UserIdMismatchError()
    
    if user_pass != 1234:
        raise UserPassMismatchError()

    print("로그인 성공!")
```

```python
# 사용자 정의 예외 한정
# () 제거 가능
def login(user_id, user_pass):
    if user_id != "user01":
        raise UserIdMismatchError
    
    if user_pass != 1234:
        raise UserPassMismatchError

    print("로그인 성공!")
```

```python
try:
    login("user02", 1234)
except (UserIdMismatchError, UserPassMismatchError) as e:
    print(e)
```
