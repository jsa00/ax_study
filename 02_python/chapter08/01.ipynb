```python
# __str__, __repr__
class Bus:                  # 클래스 정의
    # 속성 정의
    def __init__(self):     # 초기화 함수
        self.count = 0      # 승객 수
        self.passengers = []

    def ride(self, passenger):         # 탑승
        self.count += 1     # 승객 수 증가
        self.passengers.append(passenger)
        print("탑승")

    def off(self, passenger):          # 하차
        if self_count > 1:
            self.count -= 1     # 승객 수 감소
            self.passengers.remove(self.passenger)

        print("하차")

    def show_count(self):   # 현재 승객 수 현황
        print(f"승객 수: {self.count}")

    def print_self(self):   # 확인
        print(id(self))

    def __str__(self):      # __str__
        return f"승객 수: {self.count}, 승객: {self.passengers}"

class Passenger:
    def __init__(self, name, age):  # 객체를 생성하기 위해 매개변수로 정의된 지역변수
        self.name = name    # 인스턴스 변수 = 지역변수
        self.age = age

    def __str__(self):
        return f"이름: {self.name}, 나이: {self.age}"

    def __repr__(self):     # __repr__
        return f"Passenger(name='{self.name}', 'age={self.age}')"
```

```python
# 클래스와 인스턴스
bus1 = Bus()                # 클래스 호출

print(f"탑승객 수 : {bus1.count}")
bus1.ride()
```
