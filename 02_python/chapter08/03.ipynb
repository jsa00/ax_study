```python
# 클래스 변수(정적 변수), 인스턴스 변수
# 인터프리터 언어는 객체 생성 이후에도 값 할당 가능
class User:
    pass

user = User()
user.name = "고영희"
user.age = 20

print(user.name, user.age)
```

```python
# 클래스 변수(정적 변수), 인스턴스 변수
class User:
    count = 0       # 정적 변수
```

```python
user1 = User()
user2 = User()

print(user1.count, user2.count)
```

```python
user1.count = 10        

print(user1.count, user2.count)
```

```python
# 클래스 변수(정적 변수), 인스턴스 변수
# 클래스명으로 접근 가능

User.count = 100
print(User.count)
```

```python
# @classmethod, @staticmethod
class Calculator:
    count = 0

    # clz = Calculator 클래스 객체 참조 변수
    # 클래스 변수, 다른 클래스 메서드를 접근하기 위한 목적으로 사용
    @classmethod
    def add(clz, num1, num2):
        clz.count += 1
        print("호출 횟수:", clz.count)
        return num1 + num2

    @staticmethod
    def minus(num1, num2):
        return num1 - num2
```

```python
Calculator.add(10, 20)
```

```python
Calculator.minus(10, 20)
```
