```python
# 상속 정의
class Animal:
    def __init__(self, kind):
        self.kind = kind            # 동물 종류

    def move(self):
        print("동물이 이동한다")
    
class Dog(Animal):
    def __init__(self):
        super().__init__("강아지")      # 부모쪽 인스턴스 변수인 kind에 초기화

class Cat(Animal):
    def __init__(self):
        super().__init__("고양이")
```

```python
dog = Dog()
dog.move()

cat = Cat()
cat.move()
```

```python
print(dog.kind, cat.kind)
```

```python
# 메서드 오버라이딩
class Animal:
    def __init__(self, kind):
        self.kind = kind            # 동물 종류

    def move(self):
        print("동물이 이동한다")
    
class Dog(Animal):
    def __init__(self):
        super().__init__("강아지")      # 부모쪽 인스턴스 변수인 kind에 초기화

    def move(self):                     # 메서드 오버라이드
        print(self.kind + " 타닥타닥 이동한다")

class Cat(Animal):
    def __init__(self):
        super().__init__("고양이")

    def move(self):
        print(self.kind + " 사뿐사뿐 이동한다")
```

```python
dog = Dog()
dog.move()

cat = Cat()
cat.move()
```

```python
# 메서드 오버라이드
# super
class Animal:
    def __init__(self, kind):
        self.kind = kind            # 동물 종류

    def move(self):
        print("동물이 이동한다")
    
class Dog(Animal):
    def __init__(self):
        super().__init__("강아지")      # 부모쪽 인스턴스 변수인 kind에 초기화

    def move(self):                     # 메서드 오버라이드
        super().move()                  # 부모쪽 객체의 move()
        print(self.kind + " 타닥타닥 이동한다")

class Cat(Animal):
    def __init__(self):
        super().__init__("고양이")

    def move(self):
        print(self.kind + " 사뿐사뿐 이동한다")
```

```python
dog = Dog()
dog.move()

cat = Cat()
cat.move()
```

```python
# 다중 상속
class ParentA:
    def a(self):
        print("ParentA - a()")
    
    def b(self):
        print("ParentA - b()")

class ParentB:
    def a(self):
        print("ParentB - a()")
    
    def c(self):
        print("ParentB - c()")

# class Child(ParentA, ParentB):
#     pass

class Child(ParentB, ParentA):
    pass
```

```python
# 정보 은닉과 캡슐화
class Schedule:
    def __init__(self, year, month, day):
        self.year = year
        self.month = month
        self.day = day

    def __str__(self):
        return f"year={self.year}, month={self.month}, day={self.day}"
```

```python
s1 = Schedule(2026,8,18)

print(s1)
```

```python
# 값 수정 가능
# 인스턴스 변수에 직접 대입이 가능하므로 값에 대한 통제 불가!
s1.year = 2026
s1.month = 2
s1.day = 29

print(s1)
```

```python
# 프라이빗 변수
class Schedule:
    def __init__(self, year, month, day):
        self.__year = year
        self.__month = month
        self.__day = day
    
    def __str__(self):
        return f"year={self.__year}, month={self.__month}, day={self.__day}"
```

```python
s1 = Schedule(2026,8,18)

print(s1)
```

```python
# 값 변경 불가
s1.__year = 2026
s1.__month = 2
s1.__day = 29

print(s1)
```

```python
# Getter 와 Setter
class Schedule:
    def __init__(self, year, month, day):
        self.__year = year
        self.__month = month
        self.__day = day

    def get_year(self):
        return self.__year

    def get_month(self):
        return self.__month

    def get_day(self):
        return self.__day

    def set_year(self, year):
        self.__year = year

    def set_month(self, month):
        self.__month = month

    def set_day(self, day):
        if self.__month == 2 and day > 28:      # 값에 대한 통제 가능
            day = 28

        self.__day = day
    
    def __str__(self):
        return f"year={self.__year}, month={self.__month}, day={self.__day}"
```

```python
s1 = Schedule(2026,8,18)

s1.set_month(2)
s1.set_day(30)

print(s1)
```
