# 2장 4강 : 문자열 형식화
1. 1세대 : % 연산자 방식
2. 2세대 : str.format() 메서드 방식
3. 3세대 : f-string 방식

```python
# 1세대 방식
# % 연산자
message = "나의 이름은 %s이고, 나이는 %d입니다."
print(message %("철수", 20))
```

```python
# 1세대 방식
# %숫자s
message = "나의 이름은 %10s이고, 나이는 %d입니다."
print(message %("철수", 20))
```

```python
# 1세대 방식
# %대체문자숫자d
message = "주문번호: %05d" % 10
print(message)
```

```python
# 1세대 방식
# %f
message = "숫자: %f" % 10.123456789
print(message)
```

```python
# 1세대 방식
# %.nf
message = "숫자: %.2f" % 10.123456789
print(message)
```

```python
# 1세대 방식
# %010.2f
message = "숫자: %010.2f" % 10.123456789
print(message)
```

```python
# 1세대 방식
# %d
message = "숫자: %d" % 10.123456789
print(message)
```

```python
# 1세대 방식
# %s
message = "숫자: %s" % 10.123456789
print(message)
```

```python
# 2세대 방식
# str.format() 순서 지정
message = "내 이름은 {0}이고, 나이는 {1}입니다"
print(message.format("철수", 20))
```

```python
# 2세대 방식
# str.format() 순서 지정 교체
message = "내 이름은 {1}이고, 나이는 {0}입니다"
print(message.format("철수", 20))
```

```python
# 2세대 방식
# str.format() 단어 지정 교체
message = "내 이름은 {name}이고, 나이는 {age}입니다"
print(message.format(name="철수", age=20))
```

```python
# 2세대 방식
# str.format() 순서 및 단어 지정 교체
message = "내 이름은 {name}이고, 나이는 {0}입니다"
print(message.format(20, name="철수"))
```

```python
# 3세대 방식
name = "영희"
age = 20

print(f"나의 이름은 {name}이고, 나이는 {age}입니다.")
```

```python
# 3세대 방식
# 서식 지정 지시자(,)
price = 10000

print(f"가격: {price:,}")
```

```python
# 3세대 방식
# 서식 지정 지시자(:< :>)
price = 10000

print(f"가격: {price:>10,}원")
print(f"가격: {price:<10,}원")
```

```python
# 3세대 방식
# 디버깅 조절자 (=)
price = 10000

print(f"{price=:,}")
```
