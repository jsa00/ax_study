```python
# 2
# 동등성 vs 동일성
num1 = 1000
num2 = 1000

print(id(num1), id(num2))

print(num1 == num2)    # 동등성 비교
print(num1 is num2)    # 동일성 비교(주소 비교)

num3 = num1
print(num3 is num1)
```

```python
# 2
# 논리 연산자
num1 = 20

result1 = num1 > 10
result2 = num1 < 100

print(result1, result2)

# and
print(result1 and result2)

# or
print(result1 or result2)

# not
print(not (result1 and result2))
```

```python
# 2
# 논리 연산자
fruit1 = "사과"
fruit2 = "배"

result1 = fruit1 == "사과"
result2 = fruit2 == "오렌지"

print(result1, result2)
print(result1 or result2)

#
result3 = fruit1 == "사과" or fruit2 == "오렌지"
print(result3)
```

```python
# 3
# 단락 회로 평가
result = True or 10**2      # 값 자체도 참(True)
# result = True or (10**2)  # 첫 항의 결과가 True -> 결과가 이미 결정 -> 다음 연산 진행 x
print(result)

result = False or 10**2     # 첫 항의 결과가 False -> 결과가 아직 결정 x -> 다음 항 체크(연산)
print(result)

result = False and 10**2    # 첫 항의 결과가 False -> 결과가 이미 결정 -> 다음 연산 진행 x
print(result)

result = True and 10**2     # 첫 항의 결과가 Ture -> 결과가 아직 결정 x -> 다음 항 체크(연산)
print(result)
```

```python
# 3
# 복합 대입 연산자
num = 1
num += 1    # num = num + 1, 단항 연산
num += 1
num += 1

print(num)

num *= 2
print(num)
```

```python
# 5
# 부호 반전 연산자(-)
num = 100

num2 = -num     # 양수 -> 음수
print(num2)

num3 = -num2    # 음수 -> 양수
print(num3)
```

```python
# 3
# or 연산을 이용한 기본값 (없을 때 대체되는 값)
message = ""

result = message or "기본 메세지"
print(result)
```

```python
# 3
# and 연산을 이용한 간단한 조건 처리
hight = 165

result = hight >= 160 and "크다"
print(result)
```
