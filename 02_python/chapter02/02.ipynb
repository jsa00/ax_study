```python
# 2
# 명시적 형변환
num = "100"
print(type(num))

int(num) + 200

"현재 시간은 " + str(10) + "시"
```

```python
# 3
# 묵시적 형변환
num1 = 100
num2 = 3.123

print(type(num1), type(num2))

num3 = num1 + num2         # num1이 int -> float 자동 형변환(묵시적 형변환)
num4 = num1 + int(num2)     # num2를 int로 명시적 형변환 시 int로 출력

print(num3, type(num3))
print(num4, type(num4))
```

```python
# 4
# 형변환 시 주의해야 할 예외와 오류
time = "10시"
# int(time) + 1     # ValueError 발생

if time.isdigit():  # if ~.isdigit
    int(time) + 1
```

```python
# 5
# 불리언형
message = ""
message2 = "1"

if message:
    print("있음")
elif message2:
    print("있음")
else:
    print("없음")

print(type(message))

result = bool(message)
print(result, type(result))
```
