#### 4.1 데이터 로드 및 결측치, 이상치 처리

타이타닉(Titanic) 탑승객 데이터 로드 하기

- df.head() : 앞쪽 5개 조회
- df.tail() : 뒤쪽 5개 조회

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = sns.load_dataset("titanic")

# df[:5]
# df.iloc[:5]
df.head()
```

```python
# 컬럼 목록 확인
df.columns
```

데이터 요약 및 통계 확인 하기

```python
# 데이터 정보 확인
df.info()
```

```python
# 이상치 확인
df.describe()
```

결측치 파악하기

```python
# 결측치 유무 확인
df.isna().sum()
```

숫자형 데이터(나이): 전체 탑승객의 '평균(mean)'으로 빈칸 채우기

```python
# 나이 평균치 구하기
mean_age = df['age'].mean()

print("평균 나이:", mean_age)
```

```python
# 나이 결측치 평균으로 대체
df['age'] = df['age'].fillna(mean_age)

df.isna().sum()
```

문자/범주형 데이터(탑승 항구): 가장 많이 탑승한 곳인 '최빈값(mode)'으로 빈칸 채우기

```python
# 가장 많이 탑승한 곳 찾기
mode_embarked = df['embark_town'].mode()[0]

mode_embarked
```

```python
# 탑승 항구 결측치 최빈값으로 대체
df['embark_town'] = df['embark_town'].fillna(mode_embarked)
df['embarked'] = df['embark_town'].map(lambda x: x[0].upper())

df.isna().sum()
```

특정 컬럼(deck) 을 '기준'으로 좌우(열 방향) 데이터를 채우기

```python
# deck 결측치 열 방향 대체
df['deck'] = df['deck'].ffill()
df['deck'] = df['deck'].bfill()

df.isna().sum()
```

```python
df.head()
```

너무 비싼 요금(이상치) 걸러내기

```python
# 요금 평균 및 중앙값 확인
print("요금 평균:", df['fare'].mean())
print("요금 평균:", df['fare'].median())
```

```python
# 분표로 이상치 확인
plt.figure(figsize=(5,5))
plt.hist(df['fare'], bins=10)
plt.show()
```

```python
# 300 이하의 요금 걸러내기
df = df[df['fare'] <= 300]
```

결측치/이상치 보완 결과 확인

```python
df.isna().sum()
```

```python
# 300 초과 요금 확인
df[df['fare'] > 300].count()
```

#### 4.2 중복 데이터 처리하기

중복 데이터 확인하기

- duplicated() : 중복 행 확인

```python
df.duplicated().sum()
```

중복 데이터 제거하기

- drop_duplicates() : 중복 행 삭제

```python
df.drop_duplicates()[:100]      # [:100] 인덱스를 사용하여 100개 보기
```

```python
# 중복 데이터 삭제
df = df.drop_duplicates().reset_index(drop=True)

print("중복 데이터 수:", df.duplicated().sum())
```

그래프로 분석하기

```python
fig, axes = plt.subplots(1, 3, figsize=(18,5))
fig.suptitle('타이타닉 탑승객 데이터 분석', fontsize=16)  

# [히스토그램] 탑승객 나이 분포
axes[0].hist(df['age'], bins=20, color='skyblue', edgecolor='white')
axes[0].set_title("탑승객의 나이 분포")
axes[0].set_xlabel('나이')
axes[0].set_ylabel('승객수')

# [박스 플롯] 티켓 등급에 따른 요금 이상치 확인
sns.boxplot(x='class', y='fare', data=df, ax=axes[1])
axes[1].set_title('티켓 등급별 요금 분포')
axes[1].set_xlabel('티켓 등급')
axes[1].set_ylabel('티켓 요금')

# [산점도] 나이와 클리닝된 요금의 관계
scatter = axes[2].scatter(
    df['age'], 
    df['fare'], 
    c=df['survived'],       # 생존 여부(1: 생존, 0: 사망)
    cmap='coolwarm',        # 1 : 빨강, 0 : 파랑
    s=df['fare'] / 2,       # 원 사이즈 설정 (50%)
    alpha=0.7
)

fig.colorbar(scatter, ax=axes[2], label="생존 여부")  
```
