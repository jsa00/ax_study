```python
# matrix[행, 열]
print("0행 2열:", matrix[0][2], matrix[0,2])
```

```python
# matrix 지정 출력
print(matrix[0:2, 1:3])
```

```python
# matrix 0행 전체 출력
print(matrix[:, 0])
```

```python
# matrix 
print(matrix[1:, :2])
```

#### 1.2 차원 변경(reshape, flatten)과 축(Axis)

reshape(행, 열) / flatten()

- reshape(행, 열) : 1차원 배열 -> 지정된 행과 열 구조로 차원수 변경, -1 : 자동 결정
- arange(시작, 종료, 증감) -> 시작 이상 종료 미만으로 숫자 생성
- flatten() : 2차원 배열 -> 1차원으로 평탄화, 연산에 특화

```python
# reshape(행, 열)
arr = np.arange(1, 13)  # 1~12

mat_3x4 = arr.reshape(3, 4)
print(mat_3x4)
```

```python
# 자동 결정
# mat_3x4 = arr.reshape(3, -1)        
mat_3x4 = arr.reshape(-1, 4)
print(mat_3x4)
```

```python
# flatten() 
flat_arr = mat_3x4.flatten()
print(flat_arr)
```

축(axis)

2차원 배열에서 연산의 방향을 설정
- axis : 0 -> 행기준에서 연산
- axis : 1 -> 열기준에서 연산

```python
mat = np.arange(1, 7).reshape(2, -1)

print(mat, "\n")

sum_axis0 = mat.sum(axis=0)
sum_axis1 = mat.sum(axis=1)

print("행 기준 합계:", sum_axis0)
print("열 기준 합계:", sum_axis1)
```

#### 1.3 Numpy 브로드캐스팅(Broadcasting) 연산

연산을 알아서 전체 행열에 동일한 값으로 연산 실행

```python
mat = np.array([
    [10,20,30],
    [40,50,60]
])

print("곱하기", mat * 5, "\n")
print("더하기", mat + 5)
```

```python
# 각각 배열 단위로 덧셈, 곱셈 가능
vec = np.array([1,2,3])

mat + vec
```

#### 1.4 행렬 전치(Transpose) 및 행렬곱(@)

- 객체.transpose() : {행, 열} -> (열, 행) / 전치
- 객체.T : 객체.transpose() 간략 버전
- 행렬 곱 연산(@) : 

```python
# 객체.transpose()
mat = np.array([
    [1,2,3],
    [4,5,6]
])

print(mat, "\n")
print("전치 행렬:\n", mat.transpose())
```

```python
# 객체.T
mat = np.array([
    [1,2,3],
    [4,5,6]
])

print(mat, "\n")
print("전치 행렬:\n", mat.T)
```

```python
# 행렬 곱 연산(@)
mat1 = np.array([
    [10,20],
    [30,40]
])
mat2 = np.array([
    [1,2],
    [3,4]
])

mat3 = mat1 @ mat2

mat3
```
